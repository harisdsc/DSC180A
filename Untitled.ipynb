```python
# 1. Build base transaction table
Transaction_based = (
    consDF
    .drop(columns=["credit_score"])
    .merge(trxnDF, on="prism_consumer_id", how="outer")
)

Transaction_based["posted_date"] = pd.to_datetime(Transaction_based["posted_date"])
Transaction_based["year"] = Transaction_based["posted_date"].dt.year
Transaction_based["month"] = Transaction_based["posted_date"].dt.month

# 2. Keep CREDIT amounts only
Transaction_based["credit_amt"] = Transaction_based["amount"].where(
    Transaction_based["credit_or_debit"] == "CREDIT", 0
)

# 3. Monthly aggregation per consumer
monthly = (
    Transaction_based
    .groupby(["prism_consumer_id", "year", "month"], as_index=False)
    .agg(
        monthly_credit_amt=("credit_amt", "sum"),
        DQ_TARGET=("DQ_TARGET", "max")
    )
)

# 4. Per-consumer mean & std
stats = (
    monthly
    .groupby("prism_consumer_id")
    .agg(
        mean_amt=("monthly_credit_amt", "mean"),
        std_amt=("monthly_credit_amt", "std")
    )
)

# 5. Merge stats back
monthly = monthly.merge(stats, on="prism_consumer_id")

# 6. Indicator: within ±1 std (handle std = 0)
monthly["within_1_std"] = (
    (monthly["monthly_credit_amt"] >= monthly["mean_amt"] - monthly["std_amt"]) &
    (monthly["monthly_credit_amt"] <= monthly["mean_amt"] + monthly["std_amt"])
).astype(int)

monthly.loc[monthly["std_amt"] == 0, "within_1_std"] = 1

# 7. Final stability indicator per consumer
stability = (
    monthly
    .groupby("prism_consumer_id")
    .agg(
        pct_months_within_1_std=("within_1_std", "mean"),
        DQ_TARGET=("DQ_TARGET", "max")
    )
)

stability["credit_stability_flag"] = (
    stability["pct_months_within_1_std"] >= 0.8
).astype(int)

# 8. Correlation with target
stability[["credit_stability_flag", "DQ_TARGET"]].corr()
```

```python
# 1. Base table
Transaction_based = (
    consDF
    .drop(columns=["credit_score"])
    .merge(trxnDF, on="prism_consumer_id", how="outer")
)

Transaction_based["posted_date"] = pd.to_datetime(Transaction_based["posted_date"])
Transaction_based["year"] = Transaction_based["posted_date"].dt.year
Transaction_based["month"] = Transaction_based["posted_date"].dt.month

# 2. Signed transaction amount
# CREDIT = positive, DEBIT = negative
Transaction_based["signed_amt"] = np.where(
    Transaction_based["credit_or_debit"] == "CREDIT",
    Transaction_based["amount"],
    -Transaction_based["amount"]
)

# 3. Monthly net amount per consumer
monthly = (
    Transaction_based
    .groupby(["prism_consumer_id", "year", "month"], as_index=False)
    .agg(
        net_monthly_amt=("signed_amt", "sum"),
        DQ_TARGET=("DQ_TARGET", "max")
    )
)

# 4. Flag negative months
monthly["negative_month_flag"] = (monthly["net_monthly_amt"] < 0).astype(int)

# 5. Pure count of negative months
neg_count = (
    monthly
    .groupby("prism_consumer_id")
    .agg(
        neg_month_count=("negative_month_flag", "sum"),
        DQ_TARGET=("DQ_TARGET", "max")
    )
)

# 6. Amount-amplified negative months
# Sum of absolute negative net amounts
neg_amplified = (
    monthly
    .assign(neg_amt=lambda x: x["net_monthly_amt"].where(x["net_monthly_amt"] < 0, 0))
    .groupby("prism_consumer_id")
    .agg(
        neg_amt_sum=("neg_amt", lambda s: s.abs().sum())
    )
)

# 7. Combine both tests
negative_tests = neg_count.merge(neg_amplified, on="prism_consumer_id")

# 8. Correlations with target
negative_tests[["neg_month_count", "neg_amt_sum", "DQ_TARGET"]].corr()
```

```python
import pandas as pd
import numpy as np

# 1️⃣ Base table
Transaction_based = consDF.drop(columns=["credit_score"]).merge(
    trxnDF, on="prism_consumer_id", how="outer"
)

Transaction_based["posted_date"] = pd.to_datetime(Transaction_based["posted_date"])
Transaction_based["year"] = Transaction_based["posted_date"].dt.year
Transaction_based["month"] = Transaction_based["posted_date"].dt.month

# Signed amount (credit positive, debit negative)
Transaction_based["used"] = np.where(
    Transaction_based["credit_or_debit"] == "CREDIT",
    Transaction_based["amount"],
    -Transaction_based["amount"]
)

# 2️⃣ Aggregate per consumer-month-category
monthly_cat = (
    Transaction_based
    .groupby(["prism_consumer_id", "year", "month", "category"], as_index=False)
    .agg(
        used_amt=("used", "sum"),
        DQ_TARGET=("DQ_TARGET", "max")
    )
)

# 3️⃣ One-hot encode categories
cat_dummies = pd.get_dummies(monthly_cat["category"], prefix="cat")
monthly_cat = pd.concat([monthly_cat, cat_dummies], axis=1)

# 4️⃣ Aggregate to consumer-month level (presence = max of dummy)
# 1. Build dictionary for aggregation
agg_dict = {col: "max" for col in cat_dummies.columns}  # category presence
agg_dict["DQ_TARGET"] = "max"  # keep target

# 2. Aggregate
cat_features = (
    monthly_cat
    .groupby(["prism_consumer_id", "year", "month"], as_index=False)
    .agg(agg_dict)
)


# 5️⃣ Compute correlation of each category with DQ_TARGET
corrs = (
    cat_features
    .corr(numeric_only=True)["DQ_TARGET"]
    .drop("DQ_TARGET")
    .sort_values(key=lambda x: x.abs(), ascending=False)
)

# 6️⃣ Top category and correlation
top_category = corrs.index[0]
top_corr_value = corrs.iloc[0]

print("Top correlated category:", top_category)
print("Correlation with DQ_TARGET:", top_corr_value)
print("\nTop 10 category correlations:\n", corrs.head(10))
```
